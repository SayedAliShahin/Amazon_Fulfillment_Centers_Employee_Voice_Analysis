```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Amazon_Extern/youtube_cleaned.csv')
```

```python
df.columns
```

```python
df['feedback_tokens'] = df['feedback_cleaned'].apply(lambda x: str(x).split())
df[['feedback_cleaned', 'feedback_tokens']].head()
```

```python
words_to_remove = ['like', 'get', 'im']

df['feedback_tokens'] = df['feedback_tokens'].apply(lambda tokens: [word for word in tokens if word not in words_to_remove])

df[['feedback_cleaned', 'feedback_tokens']].head()
```

```python
from collections import Counter

all_words = [word for sublist in df['feedback_tokens'] for word in sublist]

word_counts = Counter(all_words)

print("Top 20 most frequent words:")
for word, count in word_counts.most_common(20):
    print(f"{word}: {count}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

all_words_updated = [word for sublist in df['feedback_tokens'] for word in sublist]
word_counts_updated = Counter(all_words_updated)

top_words = word_counts_updated.most_common(10)
words, counts = zip(*top_words)

word_freq_df = pd.DataFrame({'Word': words, 'Frequency': counts})

sns.set_theme(style="white", palette="pastel")

plt.figure(figsize=(12, 8))
# Use sns.barplot for vertical bars
ax = sns.barplot(x='Word', y='Frequency', data=word_freq_df, palette='Blues_r', hue='Word', legend=False)

# Add frequency labels on top of each bar
for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', padding=3)

plt.title('Top 10 Most Frequent Words', fontsize=16)
plt.xlabel('Word', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xticks(fontsize=10, rotation=45, ha='right') # Rotate labels for better readability
plt.yticks(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import nltk
from nltk.util import ngrams

nltk.download('punkt', quiet=True)

all_tokens = [token for sublist in df['feedback_tokens'] for token in sublist]

bigrams = list(ngrams(all_tokens, 2))

bigram_counts = Counter(bigrams)

top_bigrams = bigram_counts.most_common(10)
bigram_words = [" ".join(bigram) for bigram, count in top_bigrams]
bigram_counts_values = [count for bigram, count in top_bigrams]

bigram_freq_df = pd.DataFrame({'Bigram': bigram_words, 'Frequency': bigram_counts_values})

sns.set_theme(style="white", palette="pastel")

plt.figure(figsize=(12, 8))
ax = sns.barplot(x='Bigram', y='Frequency', data=bigram_freq_df, palette='Blues_r', hue='Bigram', legend=False)

for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', padding=3)

plt.title('Top 10 Most Frequent Bigrams ', fontsize=16)
plt.xlabel('Bigram', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xticks(fontsize=10, rotation=45, ha='right')
plt.yticks(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import nltk
from nltk.util import ngrams

# Ensure punkt tokenizer is downloaded
nltk.download('punkt', quiet=True)

# Flatten the list of tokens from the 'feedback_tokens' column
all_tokens = [token for sublist in df['feedback_tokens'] for token in sublist]

# Generate trigrams
trigrams = list(ngrams(all_tokens, 3))

# Count trigram frequencies
trigram_counts = Counter(trigrams)

# Get the top 10 most common trigrams
top_trigrams = trigram_counts.most_common(10)
trigram_words = [" ".join(trigram) for trigram, count in top_trigrams]
trigram_counts_values = [count for trigram, count in top_trigrams]

# Create a DataFrame for easier plotting
trigram_freq_df = pd.DataFrame({'Trigram': trigram_words, 'Frequency': trigram_counts_values})

# Set up the plot style, keeping it consistent with the previous charts
sns.set_theme(style="white", palette="pastel")

plt.figure(figsize=(12, 8))
# Use sns.barplot for vertical bars
ax = sns.barplot(x='Trigram', y='Frequency', data=trigram_freq_df, palette='Blues_r', hue='Trigram', legend=False)

# Add frequency labels on top of each bar
for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', padding=3)

plt.title('Top 10 Most Frequent Trigrams', fontsize=16)
plt.xlabel('Trigram', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xticks(fontsize=10, rotation=45, ha='right') # Rotate labels for better readability
plt.yticks(fontsize=10)
plt.tight_layout()
plt.show()
```
