```python
import pandas as pd
```

```python
df = pd.read_csv("/content/drive/MyDrive/Amazon_Extern/combined_sentiment_data.csv")
```

```python
df.columns
```

```python
theme_keywords = {
    'Intense Physical Demands & Shift Scheduling': ['hardships', 'exhausted', 'injured', 'High pressure ', '10 hour shifts ', 'Heavy lifting', 'stressfull', 'physical', 'walk'],
    'Competitive Payment & Benefits': ['good pay', 'good benefits ', 'balance', 'opportunity', 'growth'],
    'Management Favoritism and HR poor support': ['HR ', 'politics ', 'not appreciated', 'Management']
}
```

```python
def tag_themes(text, theme_keywords):
    matched_themes = []
    for theme, keywords in theme_keywords.items():
        if any(keyword in str(text).lower() for keyword in keywords):
            matched_themes.append(theme)
    return ', '.join(matched_themes) if matched_themes else ''
```

```python
df['Theme'] = df['text_cleaned'].apply(lambda x: tag_themes(x, theme_keywords))
```

```python
df.to_csv('Data_themes.csv', index=False)
```
