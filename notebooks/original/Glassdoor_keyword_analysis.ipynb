```python
import pandas as pd
df = pd.read_csv('/glassdoor_cleaned_processed.csv')
```

```python
noise_elements = ['get', 'job', 'like', '’', "'"]

cleaned_cols = [col for col in df.columns if col.endswith('_cleaned')]
df['merged_cleaned_text'] = df[cleaned_cols].fillna('').agg(' '.join, axis=1)

def clean_text_advanced(text):
    if not isinstance(text, str):
        return ""

    for element in noise_elements:
        text = text.replace(f" {element} ", " ")
        if text.startswith(element + " "): text = text[len(element)+1:]
        if text.endswith(" " + element): text = text[:-len(element)-1]
        text = text.replace(element, "")

    return " ".join(text.split())

df['merged_cleaned_text'] = df['merged_cleaned_text'].apply(clean_text_advanced)
```
