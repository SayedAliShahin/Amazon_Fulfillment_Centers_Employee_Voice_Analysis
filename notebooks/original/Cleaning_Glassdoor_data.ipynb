```python
import pandas as pd
```

```python
g_data = pd.read_csv('/content/glassadoor_employee_feedback_reviews - glass_reviews.csv')
```

```python
g_data.columns
```

```python
import nltk
from nltk.corpus import stopwords
import string
```

```python
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))

def cleaned_text(text):

    text = text.lower()
    text = ' '.join([word for word in str(text).split() if word not in stop_words])
    text = text.translate(str.maketrans('', '', string.punctuation))
    return text

g_data['summary_cleaned'] = g_data['summary'].fillna('').apply(cleaned_text)
display(g_data[['summary', 'summary_cleaned']].head())
```

```python
g_data['advice_to_management_cleaned'] = g_data['advice_to_management'].fillna('').apply(cleaned_text)
display(g_data[['advice_to_management', 'advice_to_management_cleaned']].head())
```

```python
g_data['review_pros_cleaned'] = g_data['review_pros'].apply(cleaned_text)
display(g_data[['review_pros', 'review_pros_cleaned']].head())
```

```python
g_data['review_cons_cleaned'] = g_data['review_cons'].apply(cleaned_text)
display(g_data[['review_cons', 'review_cons_cleaned']].head())
```

```python
g_data.to_csv('glassdoor_cleaned_processed.csv', index=False)
```

The cleaned dataset has been saved to `glassdoor_cleaned_processed.csv`.
